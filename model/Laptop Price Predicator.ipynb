```python
def add_company(inpt):
          if inpt == 'Samsung' or inpt == 'Mediacom' or inpt == 'Razer' or inpt == 'Microsoft' or inpt == 'Vero' or inpt == 'Xiaomi' or inpt == 'Chuwi' or inpt == 'Fujitsu' or inpt == 'Google' or inpt == 'LG' or inpt == 'Huawei':
              return 'other' 
          else:
              return inpt
data['Company']=data['Company'].apply(add_company)
```

```python
data['cpu_name']=data['Cpu'].apply(lambda x:" ".join(x.split()[0:3]))
```

```python
def set_processor(name):
    if name == 'Intel Core i7' or name == 'Intel Core i5' or name == 'Intel Core i3':
        return name
    else:
        if name.split()[0] == 'AMD':  # check  is processor name firt part 'AMD'?
            return 'AMD'
        else:
            return 'Other'

data['cpu_name']=data['cpu_name'].apply(set_processor)     
```

```python
def set_os(inpt):
    if inpt == 'Windows 10' or inpt == 'Windows 7' or inpt == 'Windows 10 S':
        return 'Windows'
    elif inpt == 'macOS' or inpt == 'Mac OS X':
        return 'Mac'
    elif inpt == 'Linux':
        return 'Linux'
    else:
        return 'Other'

# Apply the function
data['OpSys'] = data['OpSys'].apply(set_os)

# VERIFY THE RESULTS
print("Value counts after transformation:")
print(data['OpSys'].value_counts())

print("\nSample of original vs transformed values:")
print(data[['OpSys']].head(10))
```

```python
data = data.drop(columns=['laptop_ID','Inches','Product','ScreenResolution','Cpu','Gpu'])
```

```python
data.head(2)
```

```python
data = pd.get_dummies(data)
```

```python
data.head(2)
```

```python
data.shape
```

```python
x = data.drop('Price_euros', axis=1)
y = data['Price_euros']
```

```python
pip install scikit-learn  
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.25)
```

```python
x_train.shape,x_test.shape
```

```python
def model_acc(model):
    model.fit(x_train,y_train)
    acc = model.score(x_test,y_test)
    print(str(model)+ '⇢' +str(acc))
```

```python
from sklearn.linear_model import LinearRegression
lr = LinearRegression()
model_acc(lr)

from sklearn.linear_model import Lasso
lasso = Lasso()
model_acc(lasso)

from sklearn.tree import DecisionTreeRegressor
dt = DecisionTreeRegressor()
model_acc(dt)

from sklearn.ensemble import RandomForestRegressor
rt = RandomForestRegressor()
model_acc(rt)
```

```python
from sklearn.model_selection import GridSearchCV

parameters = {'n_estimators':[10,50,100],
              'criterion':['squared_error','absolute_error','poisson']}

grid_obj = GridSearchCV(estimator=rt, param_grid=parameters)

grid_fit = grid_obj.fit(x_train,y_train)

best_model = grid_fit.best_estimator_   #sudusuma parameters thiyena model ekak hadaganna use karai
best_model
```

```python
best_model.score(x_test,y_test)
```

```python
import pickle
with open('predictor.pickle','wb') as file:   #wb = write binary mode
    pickle.dump(best_model, file)
```
